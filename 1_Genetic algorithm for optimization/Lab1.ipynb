```python

import random

d = [
    [0, 10, 15, 20, 25],
    [10, 0, 35, 25, 17],
    [15, 35, 0, 30, 20],
    [20, 25, 30, 0, 12],
    [25, 17, 20, 12, 0]
]

n = len(d)
ps = 50
g = 200
mr = 0.1

def dist(r):
    return sum(d[r[i]][r[(i + 1) % n]] for i in range(n))

def init():
    r = list(range(n))
    random.shuffle(r)
    return r

def sel(p):
    w = [1 / dist(r) for r in p]
    return random.choices(p, weights=w, k=1)[0]

def cross(a, b):
    i, j = sorted(random.sample(range(n), 2))
    c = [None] * n
    c[i:j + 1] = a[i:j + 1]
    x = [v for v in b if v not in c]
    k = 0
    for t in range(n):
        if c[t] is None:
            c[t] = x[k]
            k += 1
    return c

def mut(r):
    r = r[:]
    if random.random() < mr:
        i, j = random.sample(range(n), 2)
        r[i], r[j] = r[j], r[i]
    return r

p = [init() for _ in range(ps)]
best = min(p, key=dist)

for _ in range(g):
    p.sort(key=dist)
    q = [p[0][:]]

    while len(q) < ps:
        a = sel(p)
        b = sel(p)
        c = mut(cross(a, b))
        q.append(c)

    p = q
    x = min(p, key=dist)

    if dist(x) < dist(best):
        best = x[:]

r = [x + 1 for x in best]
r.append(r[0])

print("Best Route:", " -> ".join(map(str, r)))
print("Minimum Distance:", dist(best))
```
